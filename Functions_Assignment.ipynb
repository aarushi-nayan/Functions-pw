Q6.  Explain the concept of generators in Python and how they are defined.

---> Generators are special functions in Python that return values one at a time using the yield keyword instead of returning all at once like return.

--Key Points

Generators produce values lazily — only when needed.

They help save memory and improve performance when working with large data.

They automatically create an iterator.

```python
#"Example"
def count_up_to(n):
    i = 1
    while i <= n:
        yield i  # yields one value at a time
        i += 1

# Using the generator
for num in count_up_to(5):
    print(num)
```

How is it different from return?

return ends the function immediately.

yield pauses the function, saves its state, and resumes when called again.


Q7. What are the advantages of using generators over regular functions?

---> 1. Memory Efficiency
Generators don’t store all values in memory at once.
They yield one value at a time, making them perfect for large data or infinite loops.

Generators produce values on the fly, only when asked, saving time and resources.

3. Cleaner Code
No need to manage the state or use complex loops.

4. Infinite Sequences
They can generate endless data without crashing the program.

5. Improved Performance
Since they return data one piece at a time, performance is better for long processes like reading big files or processing streams.

Q10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

---> These three functions are used for functional programming and can be applied to iterables (like lists). Here’s a breakdown of how each function works and when to use them:

1. map():
Purpose: Applies a function to every element in the iterable and returns a new iterable with the results.

Usage: When you want to transform or modify all elements in an iterable.



```python
# Example: Doubling each element in the list
nums = [1, 2, 3, 4]
doubled = map(lambda x: x * 2, nums)
print(list(doubled))  # Output: [2, 4, 6, 8]
```

2. reduce():
Purpose: Reduces an iterable to a single value by applying a binary function cumulatively to the elements. This function is available in the functools module.

Usage: When you want to aggregate or combine values from an iterable into a single result (like summing all numbers, finding a product, etc.).

```python
from functools import reduce

# Example: Finding the product of all elements
nums = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, nums)
print(product)  # Output: 24
```

3. filter():
Purpose: Filters the elements of an iterable based on a condition (predicate) and returns an iterable with only the elements that satisfy that condition.

Usage: When you want to filter out elements based on a certain condition.

```python
# Example: Filtering even numbers
nums = [1, 2, 3, 4, 5, 6]
even_nums = filter(lambda x: x % 2 == 0, nums)
print(list(even_nums))  # Output: [2, 4, 6]
```

map(): Modify each item.

reduce(): Combine items into a single value.

filter(): Select items based on a condition.

Q1.  Write a Python function that takes a list of numbers as input and
     returns the sum of all even numbers in the list.

```python
def sum_of_even_numbers(numbers):
    even_sum = 0
    for num in numbers:
        if num % 2 == 0:
            even_sum += num
    return even_sum

# Example usage:
nums = [10, 2, 8, 3, 2]
result = sum_of_even_numbers(nums)
print("Sum of even numbers:", result)
```

Q4. Write a Python function that checks if a given number is prime or not from 1 to 200.

--->

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Create a list to store prime numbers
prime_list = []

# Check and store prime numbers from 1 to 200
for number in range(1, 201):
    if is_prime(number):
        prime_list.append(number)

# Print the list of prime numbers
print("Prime numbers from 1 to 200:")
print(prime_list)
```

Q5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms

--->

```python
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return self.a
        elif self.count == 1:
            self.count += 1
            return self.b
        else:
            next_fib = self.a + self.b
            self.a, self.b = self.b, next_fib
            self.count += 1
            return next_fib

# Example usage:
fib = FibonacciIterator(10)  # Generate first 10 terms
for num in fib:
    print(num)
```

Q6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

--->


```python
def powers_of_two(exponent):
    for i in range(exponent + 1):
        yield 2 ** i

# Example usage:
for power in powers_of_two(5):  # Powers of 2 up to 2^5
    print(power)
```

Q7.  Implement a generator function that reads a file line by line and yields each line as a string.



```python
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line  # Yield each line as it is

# Example usage:
file_path = 'example.txt'  # Replace with your file path
for line in read_file_line_by_line(file_path):
    print(line, end='')  # 'end' is used to avoid double newlines
```

Q9.  Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
# List of temperatures in Celsius
celsius_temps = [0, 20, 30, 40, 100]

# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# Use map() to convert the temperatures
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))

# Print the result
print(fahrenheit_temps)
```

Q10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

--->

```python
# Function to check if a character is a vowel
def is_not_vowel(char):
    return char.lower() not in 'aeiou'

# Given string
input_string = "Hello, World!"

# Use filter() to remove vowels
result = ''.join(filter(is_not_vowel, input_string))

# Print the result
print(result)
```
